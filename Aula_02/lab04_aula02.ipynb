```python
# ============================================================
# LAB 04 - AULA 02 (MLCB): DESAFIO NLU PARA AGÊNCIA DE VIAGENS
# ============================================================

import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split


# 1. DATASET PRÓPRIO - INTENÇÕES DE UMA AGÊNCIA DE VIAGENS
dados_viagens = {
    'mensagem': [
        'Quero comprar uma passagem para São Paulo',
        'Gostaria de comprar uma passagem aérea',
        'Preciso reservar um voo para o Rio de Janeiro',
        'Quero adquirir uma passagem para Salvador',
        'Quero cancelar minha reserva',
        'Preciso cancelar minha passagem',
        'Como faço para cancelar o voo?',
        'Gostaria de cancelar minha reserva de hotel',
        'Quero falar com um atendente',
        'Preciso conversar com uma pessoa',
        'Gostaria de falar com o atendimento',
        'Pode me encaminhar para um atendente?'
    ],

    'intencao': [
        'comprar_passagem','comprar_passagem','comprar_passagem','comprar_passagem',
        'cancelar_reserva','cancelar_reserva','cancelar_reserva','cancelar_reserva',
        'falar_atendente','falar_atendente',
        'falar_atendente','falar_atendente'
    ]
}

df = pd.DataFrame(dados_viagens)

print("--- DATASET ---")
print(df)
print()


# 2. SEPARAÇÃO DOS ATRIBUTOS (X) E RÓTULOS (Y)
X = df['mensagem']
y = df['intencao']



# 3. DIVISÃO DO DATASET EM TREINO E TESTE
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)


# 4. VETORIZAÇÃO DOS TEXTOS - TF-IDF
tfidf = TfidfVectorizer()

X_train_vec = tfidf.fit_transform(X_train)
X_test_vec = tfidf.transform(X_test)



# 5. TREINAMENTO DO MODELO - REGRESSÃO LOGÍSTICA
modelo = LogisticRegression()

modelo.fit(X_train_vec, y_train)

# 6. TESTE DO MODELO
predicoes_teste = modelo.predict(X_test_vec)

print("--- RESULTADOS DO TESTE ---")

for frase, intencao_real, intencao_predita in zip(
    X_test,
    y_test,
    predicoes_teste
):
    print(f"Mensagem: '{frase}'")
    print(f"Intenção real: [{intencao_real}]")
    print(f"Intenção predita: [{intencao_predita}]")
    print()


# 7. TESTE COM 3 FRASES INÉDITAS
novas_frases = [
    "Gostaria de adquirir um voo para Recife",
    "Preciso desistir da minha reserva",
    "Quero conversar com alguém do atendimento"
]

novas_frases_vec = tfidf.transform(novas_frases)

predicoes = modelo.predict(novas_frases_vec)


# 8. EXIBIÇÃO DAS PREDIÇÕES
print("--- RESULTADOS DO LAB 04 ---")

for frase, intencao in zip(novas_frases, predicoes):
    print(
        f"Mensagem: '{frase}' ==> "
        f"Intenção Predita: [{intencao}]"
    )
```
