```python
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import nltk
import spacy

from nltk.corpus import stopwords
from gensim.models import FastText
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

# Preparar ambiente
nltk.download("stopwords", quiet=True)
stop_words_pt = set(stopwords.words("portuguese"))
nlp = spacy.load("pt_core_news_sm")

print("Ambiente preparado com sucesso.")



def limpar_e_lemmatizar(texto):
    texto_limpo = texto.lower()

    texto_limpo = re.sub(
        r'[^a-záàâãéèêíïóôõöúçñ\s]',
        '',
        texto_limpo
    )

    doc = nlp(texto_limpo)

    tokens_filtrados = []

    for token in doc:


        if token.is_space:
            continue

        if token.text in stop_words_pt:
            continue


        if len(token.text) <= 1:
            continue

        tokens_filtrados.append(token.lemma_)

    return " ".join(tokens_filtrados)


frase_teste = (
    "Gostaria de saber se vocês estão DEVOLVENDO "
    "os valores das mesas compradas!!!"
)

resultado = limpar_e_lemmatizar(frase_teste)

print("\nFrase original:")
print(frase_teste)

print("\nFrase processada:")
print(resultado)


def analisar_preprocessamento(texto):

    texto_original = texto

    texto_limpo = texto.lower()

    texto_limpo = re.sub(
        r'[^a-záàâãéèêíïóôõöúçñ\s]',
        '',
        texto_limpo
    )

    doc = nlp(texto_limpo)

    tokens_antes = [
        token.text
        for token in doc
        if not token.is_space
    ]

    tokens_removidos = []
    tokens_finais = []

    for token in doc:

        if token.is_space:
            continue


        if token.text in stop_words_pt:
            tokens_removidos.append(token.text)
            continue

        if len(token.text) <= 1:
            tokens_removidos.append(token.text)
            continue

        tokens_finais.append(token.lemma_)

    texto_normalizado = " ".join(tokens_finais)

    print("=" * 60)

    print("Texto original:")
    print(texto_original)

    print("\nTexto normalizado:")
    print(texto_normalizado)

    print("\nQuantidade de caracteres:")
    print(len(texto_original))

    print("\nQuantidade de tokens antes:")
    print(len(tokens_antes))

    print("\nQuantidade de tokens depois:")
    print(len(tokens_finais))

    print("\nTokens removidos:")
    print(tokens_removidos)

    print("\nTokens finais:")
    print(tokens_finais)

    print("=" * 60)


analisar_preprocessamento(
    "MEU sofá!!! chegou quebrado e quero DEVOLVER!!!"
)
```

```python

df = pd.read_csv("sac_moveis_ac2_treino.csv")

print("Dataset carregado com sucesso.")
print(df.head())

df["mensagem_processada"] = df["mensagem"].apply(
    limpar_e_lemmatizar
)

print(
    df[
        [
            "mensagem",
            "mensagem_processada",
            "intencao"
        ]
    ].head()
)

corpus_tokenizado = (
    df["mensagem_processada"]
    .str.split()
    .tolist()
)

print(corpus_tokenizado[:5])

modelo_fasttext = FastText(
    sentences=corpus_tokenizado,
    vector_size=50,
    window=3,
    min_count=1,
    workers=4,
    sg=1
)

print("Modelo FastText treinado com sucesso.")

vetor_sofa = modelo_fasttext.wv["sofá"]


def obter_vetor_frase(frase, modelo):

    palavras = frase.split()

    vetores = []

    for palavra in palavras:

        vetor = modelo.wv[palavra]

        vetores.append(vetor)

    if len(vetores) == 0:

        return np.zeros(
            modelo.vector_size
        )

    vetor_medio = np.mean(
        vetores,
        axis=0
    )

    return vetor_medio


frase_teste = limpar_e_lemmatizar(
    "Quero devolver meu sofá"
)

print("Frase original:")
print("Quero devolver meu sofá")

print("Frase processada:")
print(frase_teste)

vetor_frase = obter_vetor_frase(
    frase_teste,
    modelo_fasttext
)


print("Dimensão do vetor:")
print(vetor_frase.shape)

X_vetores = np.array(
    [
        obter_vetor_frase(
            frase,
            modelo_fasttext
        )
        for frase in df["mensagem_processada"]
    ]
)

y = df["intencao"].values

print("Formato de X_vetores:")
print(X_vetores.shape)

print("Quantidade de intenções:")
print(len(y))

print("Mensagem:")
print(df["mensagem"].iloc[0])

print("Mensagem processada:")
print(df["mensagem_processada"].iloc[0])

print("Intenção:")
print(y[0])


print("Quantidade de dimensões:")
print(len(X_vetores[0]))
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_vetores,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)


modelo = LogisticRegression(
    max_iter=1000
)

modelo.fit(
    X_train,
    y_train
)


y_pred = modelo.predict(
    X_test
)

print(
    classification_report(
        y_test,
        y_pred
    )
)


def classificar_mensagem(
    mensagem,
    modelo,
    modelo_embedding,
    limiar=0.50
):

    mensagem_processada = limpar_e_lemmatizar(
        mensagem
    )

    vetor = obter_vetor_frase(
        mensagem_processada,
        modelo_embedding
    )

    vetor = vetor.reshape(
        1,
        -1
    )

    probabilidades = modelo.predict_proba(
        vetor
    )[0]

    indice_maior = np.argmax(
        probabilidades
    )

    confianca = probabilidades[
        indice_maior
    ]

    intencao = modelo.classes_[
        indice_maior
    ]

    if confianca >= limiar:

        return (
            intencao,
            confianca
        )

    else:

        return (
            "FALLBACK_HUMANO",
            confianca
        )


testes = [
    "quero devolver meu sofá",
    "como faço para realizar a devolução?",
    "cadê meu pedido?",
    "meu pedido nao chego",
    "qual é a previsão do tempo?"
]


for mensagem in testes:

    resultado, confianca = classificar_mensagem(
        mensagem,
        modelo,
        modelo_fasttext,
        limiar=0.50
    )

    print("\nMensagem:")
    print(mensagem)

    print("Resultado:")
    print(resultado)

    print("Confiança:")
    print(
        f"{confianca:.2%}"
    )
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

modelo_logistico = LogisticRegression(
    max_iter=1000
)

modelo_knn = KNeighborsClassifier(
    n_neighbors=3
)

modelo_logistico.fit(
    X_train,
    y_train
)

modelo_knn.fit(
    X_train,
    y_train
)

y_pred_logistico = modelo_logistico.predict(
    X_test
)

y_pred_knn = modelo_knn.predict(
    X_test
)

accuracy_logistico = accuracy_score(
    y_test,
    y_pred_logistico
)

precision_logistico = precision_score(
    y_test,
    y_pred_logistico,
    average="weighted"
)

recall_logistico = recall_score(
    y_test,
    y_pred_logistico,
    average="weighted"
)

f1_logistico = f1_score(
    y_test,
    y_pred_logistico,
    average="weighted"
)

accuracy_knn = accuracy_score(
    y_test,
    y_pred_knn
)

precision_knn = precision_score(
    y_test,
    y_pred_knn,
    average="weighted"
)

recall_knn = recall_score(
    y_test,
    y_pred_knn,
    average="weighted"
)

f1_knn = f1_score(
    y_test,
    y_pred_knn,
    average="weighted"
)

print("Regressão Logística")
print(f"Accuracy:  {accuracy_logistico:.2%}")
print(f"Precision: {precision_logistico:.2%}")
print(f"Recall:    {recall_logistico:.2%}")
print(f"F1-Score:  {f1_logistico:.2%}")

print("\nKNN")
print(f"Accuracy:  {accuracy_knn:.2%}")
print(f"Precision: {precision_knn:.2%}")
print(f"Recall:    {recall_knn:.2%}")
print(f"F1-Score:  {f1_knn:.2%}")

resultado_modelos = pd.DataFrame({
    "Modelo": [
        "Regressão Logística",
        "KNN"
    ],
    "Accuracy": [
        accuracy_logistico,
        accuracy_knn
    ],
    "Precision": [
        precision_logistico,
        precision_knn
    ],
    "Recall": [
        recall_logistico,
        recall_knn
    ],
    "F1": [
        f1_logistico,
        f1_knn
    ]
})

resultado_modelos["Accuracy"] = (
    resultado_modelos["Accuracy"] * 100
).round(2)

resultado_modelos["Precision"] = (
    resultado_modelos["Precision"] * 100
).round(2)

resultado_modelos["Recall"] = (
    resultado_modelos["Recall"] * 100
).round(2)

resultado_modelos["F1"] = (
    resultado_modelos["F1"] * 100
).round(2)

print("\nTabela comparativa:")
print(resultado_modelos)

```
